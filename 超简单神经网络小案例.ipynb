```python
import torch
import torch.nn as nn
```

```python
class ModelDemo(nn.Module):
    #1、在init魔法方法中，完成初始化
    def __init__(self):
        #初始化父类
        super().__init__()
        #搭建神经网络，隐藏层+输入层
        #1、隐藏层1，输入特征数3，输出特征3
        self.linear1 = nn.Linear(in_features = 3,out_features = 3)
        #
        self.linear2 = nn.Linear(in_features = 3,out_features=2)
        #
        self.linear3 = nn.Linear(in_features = 2,out_features = 2)
        #对隐藏层进行参数初始化
        nn.init.xavier_normal_(self.linear1.weight)
        nn.init.zeros_(self.linear1.bias)
        #第二层
        nn.init.kaiming_normal_(self.linear2.weight)
        nn.init.zeros_(self.linear2.bias)
    #前向传播：输入 ---隐藏---输出
    def forward(self,x):
        #第一层：隐藏层计算：加权求和+激活函数
        x = self.linear1(x)
        x = torch.sigmoid(x)  #激活
        #第二层
        x = self.linear2(x)
        x = torch.relu(x)
        #第三层,dim = -1代表按行计算
        x = self.linear3(x)
        x = torch.softmax(x,dim=-1)
        #返回预测值
        return x
        
        
        
```

```python
#模型训练
def train():
    #创建模型对象：
    my_module = ModelDemo()
    #创建数据集样本，随机生成
    data = torch.randn(size = (5,3))
    print(f'data:{data}')
    print(f'data.shape:{data.shape}')
    print(f'{data.requires_grad}')  #False
    #调用神经网络模型进行模型训练
    output = my_module(data)   #底层自动调用了forward方法进行了前向传播
    print(output)
    print(output.shape)
train()
```
