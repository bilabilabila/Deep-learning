```python
#案例：演示词嵌入层API
#RNN:全称循环神经网络，主要处理序列数据
#序列数据：后边数据对前边数据有依赖，例如：天气预测，股市分析，文本生成
#组成：词嵌入层、循环网络层、输出层
# 词嵌入层：把词（词对应的索引）转为词向量
```

```python
import torch
import jieba
import torch.nn as nn
```

```python
#1、定义函数，用于演示词嵌入的API
def dm01():
    text = '北京东奥的进度条已经过半，不少外国运动员在完成自己的比赛后踏上归途。'
    words = jieba.lcut(text)
    print(words)
    #创建词嵌入层,参1：词表大小，参2：词向量的维度
    embed = nn.Embedding(len(words),embedding_dim=4)
    #获取每个词对象的下标索引
    for i,word in enumerate(words):
        #print(i,word)
        #把词索引（张量形式）转换为词向量
        word2vec = embed(torch.tensor(i)) #随机的每次都不一样
        print(f'词:{word},\t\t词向量：{word2vec}')
dm01()
```

```python
#演示RNN层（循环网络层）的API
#循环网络层的作用：基于上一次的隐藏状态+本次的输入 → 本次的隐藏状态 + 本次的输出
#公式 本次的隐藏状态 = tanh（上次的隐藏状态加权求和 + 本次的输入加权求和）
# 本次的输出 = 本次的隐藏状态加权求和，有词汇表中所有词的概率，选概率最大的那个词作为最终预测结果
#简单总结RNN： 词嵌入层（将词转换为词向量表示） RNN：逐步处理词向量
```

```python
import torch
import torch.nn as nn
```

```python
#大白话：RNN就像你的大脑，在看电影的过程中，记住剧情
#创建循环网络层
#参1：词向量维度：就像电影里每一帧画面有128个细节（颜色，动作，表情），RNN（你的大脑）每次看到的都是这128个细节
#参2：隐藏状态向量维度，你的大脑能记住的剧情信息量有多大，比如说：记住角色关系，前景提要，值越大，能存储的剧情记忆就越多。
#参3：隐藏层数量，默认是1，大脑只有1层来处理，如果是2，说明有2层大脑，第1层处理，第2层深加工。
rnn = nn.RNN(input_size = 128,hidden_size = 256,num_layers = 1)

#定义变量：
#参1、每个句子词的个数 （长度），假设电影总共有5帧画面
#参2、句子的数量   同时有32个人在看这5帧画面
#参3、词向量的维度  每一帧画面的细节是128个。
x = torch.randn(size = (5,32,128))
#定义变量：记录：上一时刻的隐藏状态
#参1：隐藏层的层数，和RNN的num_layer一样，一层大脑
#参2，句子的数量，批量大小，和x的参2一样，32个人在看电影
#参3 隐藏层的状态向量维度，和RNN的hidden_size一样。256
h0 = torch.randn(size = (1,32,256))

#调用RNN处理
#参1：本次的输入，每个时间步的输出，包含了所有时间步的隐藏状态，大脑能记住的剧情 → 共5帧，32个人，256维度。
#参2：上一次的隐藏状态，最后一个时间步的隐藏状态  看完最后1帧后，大脑里的最新剧情记忆。 → 只关注最后一帧。
output,h1 = rnn(x,h0)
print(output.shape)
print(h1.shape)
```
